```python
import pandas as pd
import numpy as np
from pathlib import Path

print("Libraries imported successfully!")
```

```python
data = {
    "Student_ID": list(range(101, 121)),
    "Name": [
        "Aarav", "Ishwari", "Rupali", "Sneha", "Rahul",
        "Priya", "Amit", "Neha", "Rohan", "Pooja",
        "Karan", "Anjali", "Vikas", "Sakshi", "Om",
        "Kavya", "Aditya", "Meera", "Sahil", "Riya"
    ],
    "Department": ["IT"] * 20,
    "Python": [85, 92, 76, 65, 45, 88, 72, 95, 58, 81, 67, 90, 74, 83, 39, 96, 62, 79, 55, 87],
    "NumPy": [80, 95, 72, 68, 48, 84, 75, 91, 60, 78, 69, 93, 71, 85, 42, 94, 65, 76, 57, 89],
    "Pandas": [82, 90, 78, 63, 40, 86, 70, 94, 55, 83, 64, 89, 77, 81, 38, 97, 61, 80, 52, 85],
    "Data_Analysis": [88, 94, 74, 70, 46, 89, 73, 92, 62, 80, 68, 91, 75, 84, 41, 95, 66, 78, 59, 86],
    "Attendance": [90, 95, 82, 76, 68, 92, 80, 98, 72, 85, 78, 96, 88, 91, 60, 97, 74, 83, 70, 94]
}

students = pd.DataFrame(data)

students.to_csv("students.csv", index=False)

print("Dataset created successfully!")
print("Total students:", len(students))

display(students.head())
```

```python
print("Dataset shape:", students.shape)

print("\nColumn names:")
print(students.columns.tolist())

print("\nMissing values:")
display(students.isnull().sum())

print("\nDataset information:")
students.info()
```

```python
SUBJECT_COLUMNS = [
    "Python",
    "NumPy",
    "Pandas",
    "Data_Analysis"
]

REQUIRED_COLUMNS = [
    "Student_ID", "Name", "Department",
    *SUBJECT_COLUMNS, "Attendance"
]

missing_columns = [
    column for column in REQUIRED_COLUMNS
    if column not in students.columns
]

if missing_columns:
    raise ValueError(f"Missing columns: {missing_columns}")

if students.empty:
    raise ValueError("Dataset is empty.")

if students[REQUIRED_COLUMNS].isnull().any().any():
    raise ValueError("Dataset contains missing values.")

if not students[SUBJECT_COLUMNS].apply(
    lambda column: column.between(0, 100).all()
).all():
    raise ValueError("Marks must be between 0 and 100.")

if not students["Attendance"].between(0, 100).all():
    raise ValueError("Attendance must be between 0 and 100.")

print("All data validation checks passed!")
```

```python
def assign_grade(average):
    if average >= 90:
        return "A+"
    elif average >= 80:
        return "A"
    elif average >= 70:
        return "B"
    elif average >= 60:
        return "C"
    elif average >= 50:
        return "D"
    else:
        return "F"


print("Grading function created successfully!")
```

```python
def calculate_student_results(data):
    results = data.copy()

    results["Total_Marks"] = results[SUBJECT_COLUMNS].sum(axis=1)

    results["Average_Marks"] = np.round(
        results[SUBJECT_COLUMNS].mean(axis=1), 2
    )

    results["Grade"] = results["Average_Marks"].apply(assign_grade)

    results["Result"] = np.where(
        (results[SUBJECT_COLUMNS] >= 35).all(axis=1),
        "Pass",
        "Fail"
    )

    results["Rank"] = results["Average_Marks"].rank(
        method="min",
        ascending=False
    ).astype(int)

    return results


results = calculate_student_results(students)

display(results.head(10))
```

```python
display_columns = [
    "Student_ID",
    "Name",
    "Department",
    "Total_Marks",
    "Average_Marks",
    "Grade",
    "Result",
    "Attendance",
    "Rank"
]

display(
    results[display_columns]
    .sort_values(["Rank", "Name"])
    .reset_index(drop=True)
)
```

```python
total_students = len(results)

passed = int((results["Result"] == "Pass").sum())
failed = int((results["Result"] == "Fail").sum())

class_average = round(
    float(np.mean(results["Average_Marks"])), 2
)

highest_average = results["Average_Marks"].max()
lowest_average = results["Average_Marks"].min()

pass_percentage = round(
    (passed / total_students) * 100, 2
)

print("========== CLASS PERFORMANCE SUMMARY ==========")
print("Total students:", total_students)
print("Class average:", class_average)
print("Highest average:", highest_average)
print("Lowest average:", lowest_average)
print("Students passed:", passed)
print("Students failed:", failed)
print("Pass percentage:", pass_percentage, "%")
```

```python
subject_rows = []

for subject in SUBJECT_COLUMNS:
    subject_rows.append({
        "Subject": subject.replace("_", " "),
        "Average_Marks": round(
            float(np.mean(students[subject])), 2
        ),
        "Highest_Marks": students[subject].max(),
        "Lowest_Marks": students[subject].min()
    })

subject_summary = pd.DataFrame(subject_rows)

display(subject_summary)

best_subject = subject_summary.loc[
    subject_summary["Average_Marks"].idxmax()
]

weakest_subject = subject_summary.loc[
    subject_summary["Average_Marks"].idxmin()
]

print("Best subject:", best_subject["Subject"])
print("Lowest-average subject:", weakest_subject["Subject"])
```

```python
top_students = results.sort_values(
    ["Average_Marks", "Name"],
    ascending=[False, True]
).head(3)

print("========== TOP THREE STUDENTS ==========")

display(
    top_students[
        ["Rank", "Student_ID", "Name", "Average_Marks", "Grade"]
    ]
)
```

```python
low_attendance = results[results["Attendance"] < 75]

print("========== ATTENDANCE ANALYSIS ==========")

if low_attendance.empty:
    print("All students have attendance of 75% or above.")
else:
    display(
        low_attendance[
            ["Student_ID", "Name", "Department", "Attendance"]
        ]
    )
```

```python
OUTPUT_FILE = Path("student_analysis_output.csv")

results.to_csv(OUTPUT_FILE, index=False)

print("Results saved successfully!")
print("File location:", OUTPUT_FILE.resolve())

display(pd.read_csv(OUTPUT_FILE).head())
```
